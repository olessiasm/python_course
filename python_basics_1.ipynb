Python умеет сравнивать объекты, и для этого существуют операторы, такие
как меньше, больше, равно и не равно. Результатом любой операции
сравнения будет логическое (булевое) значение: True («Истина») или False
(«Ложь»).
|Операторы|Значения|Выражения|
|:---:|:---:|:---:|
|>|Больше|a>b|
|<|Меньше|a<b|
|==|Равно|a==b|
|>=|больше или равно|a>=b|
|<=|меньше или равно|a<=b|
|!=| не равно|a!=b|

```python
income_2023 = 100
income_2024 = 105.5
income_2025 = 105.5
print(income_2023 == income_2024)
print(income_2024 == income_2025)
```

```python
comparison_1 = income_2023 >= income_2024
comparison_2 = income_2024 >= income_2025
print(comparison_1)
print(comparison_2)
```

Операции сравнения можно записывать в цепочку. Результатом цепочки
будет True, если все сравнения истинны. Если хотя бы одно из сравнений
ложно, то результат будет False. Например:


```python
print(income_2023 < income_2024 < income_2025)
```

##Сравнение строк
При сравнении строк Python использует их числовое представление в таблице
символов Unicode. Такой метод сравнения называется лексиграфическим.
Посмотреть, какой порядковый номер в таблице Unicode скрывается за
символом можно с помощью функции ord().

```python
print(ord('L'))
print(ord('l'))
print('L' >= 'l')
```

При сравнении строк, состоящих из нескольких символов, сначала

будет получен результат сравнения. Если первые символы равны, то
сравниваются вторые символы, и так далее.
В этом примере результат True, потому что буква «О» имеет больший
порядковый номер, чем «E»

```python
name_1 = 'Oleg'
name_2 = 'Olga'
print(name_1 > name_2)
```

#Приоритет операторов
Важно помнить о приоритете операторов. Оператор not имеет наивысший
приоритет, затем идёт and, а потом or. Это может влиять на результат
выражений, поэтому следите за порядком выполнения.

```python
print(True and True or True and False or not True)
print(True or False or False)
# print(True or False)
# print(True)
```

В этом примере сначала выполнится операция not True (результат: False),
следом выполнится крайнее слева True and True (результат: True), далее
True and False (результат: False). Промежуточный результат: True or
False or False. Здесь в первую очередь проверяем первые True or
False (результат: True), получаем промежуточный результат: True or
False. Итоговый результат этого выражения: True.
Если добавить скобки, ситуация изменится:

```python
print(True and (True or True) and False or not True)
# print(True and True and False or False)
# print(True and False or False)
# print(False or False)
# print(False)
```

##Ключевое слово else
Условная конструкция может быть дополнена блоком else, который
выполнится, если условие ложно. Использование этого блока опционально.
Блок else сработает, если не сработали предыдущие проверки. После блока
else никакие уловия не задаются, потому что он подразумевает, что действие в
этом блоке будет выполнено «во всех остальных случаях».
Общий синтаксис:
if Условие: # если это условие истинно,
Блок инструкций 1 # выполни эти инструкции
else: # во всех остальных случаях
Блок инструкций 2 # выполни эти инструкции

```python
is_rainy = False
if is_rainy:
    print('Возьми зонт')
else:
    print('Не бери зонт')
```

###Тернарный условный оператор
Если у нас есть простая условная конструкция c одним блоком if и одним
блоком else, и в каждом из этих блоков по одной строке, мы можем
использовать тернарный оператор для записи в одну строку. Это позволяет
сделать код более компактным, но иногда менее читаемым.
Общий синтаксис:
value_if_true if condition else value_if_false


```python
num = 5
if num % 2 == 0:
    print('Чётное')
else:
    print('Нечётное')
```

```python
num = 5
print('Чётное')if num % 2 == 0 else print('Нечётное')
```

##Ключевое слово elif
**elif** — это сокращение от английского «else if» (иначе если). Блок elif
позволяет задавать дополнительные условия, если блока if недостаточно. Он
всегда используется после блока if и до блока else. Блоков elif может быть
сколько угодно.


Так же как в случае с блоком if, блок elif должен содержать условие,
результатом которого является «Истина» (True) или «Ложь» (False).

Пример:

Предположим такую условную ситуацию: у нас есть возраст выпускника (age) и
оценка выпускника за какой-то экзамен (grade). Мы можем принять
выпускника в учебное заведение только после 18 лет. Если выпускнику
больше 18 лет, то делаем проверку баллов: если баллов меньше 100, то
отправляем выпускника на пересдачу; если баллов меньше 150, то — в
колледж; а если количество баллов меньше либо равно 200, то — в МГУ.
Если выпускнику меньше 18 лет, то ждём его совершеннолетия.#

```python
age = 20
grade = 90
if age >= 18:
    if grade < 100:
        print('На тяжелую и низкооплачиваемую работу')
    elif grade < 170:
        print('В колледж')
    elif grade <= 220:
        print('В Саратов')
    elif grade <= 300:
        print('В МГУ')
else:
    print('Подождём совершеннолетия')
```
