## 🛠 Утилита для автопроверок

Следующая ячейка определяет вспомогательные функции. **Запустите её один раз перед началом практики.**

````python
# Утилиты для автоматической проверки заданий.
# Запустите эту ячейку один раз.

import io
import contextlib


def check_value(name, actual, expected):
    """Проверяет, что переменная равна ожидаемому значению."""
    if actual == expected:
        print(f"✅ Задание пройдено: {name} = {actual!r}")
        return True
    else:
        print(f"❌ Ошибка: ожидалось {name} = {expected!r}, получено {actual!r}")
        return False


def check_output(student_func, expected_output):
    """Запускает student_func(), перехватывает stdout и сравнивает с ожидаемым."""
    buffer = io.StringIO()
    try:
        with contextlib.redirect_stdout(buffer):
            student_func()
    except Exception as e:
        print(f"❌ Ошибка при выполнении: {type(e).__name__}: {e}")
        return False

    actual = buffer.getvalue().strip()
    expected = expected_output.strip()

    if actual == expected:
        print("✅ Задание пройдено")
        return True
    else:
        print("❌ Вывод не совпадает с ожидаемым.")
        print("--- Ожидалось ---")
        print(expected)
        print("--- Получено ---")
        print(actual)
        return False


print("Утилиты автопроверки загружены.")
````

# Часть 1. Словари (`dict`)

## 1.1. Что такое словарь

**Словарь** (`dict`) — это изменяемая коллекция пар **«ключ → значение»**. Словари используются тогда, когда данные удобно связать: например, слово и его перевод, студент и его оценки, буква и её частота в тексте.

В отличие от списка, где доступ идёт по числовому индексу, в словаре доступ идёт по **ключу** — это может быть строка, число или другой неизменяемый объект.

### 1.1.1. Синтаксис

    d = {"ключ1": значение1, "ключ2": значение2}

Примеры:

    english_to_russian = {"hello": "привет", "cat": "кот", "book": "книга"}
    student_scores = {"Anna": 85, "Ivan": 72, "Olga": 91}
    empty = {}

### 1.1.2. Ключевые свойства словаря

| Свойство | Описание |
|----------|----------|
| Изменяемость | Можно добавлять, удалять, менять элементы |
| Уникальность ключей | Два одинаковых ключа невозможны — второй перезапишет первый |
| Порядок | С Python 3.7 сохраняется порядок добавления |
| Типы ключей | Ключом может быть любой **неизменяемый** объект: строка, число, кортеж |
| Типы значений | Значением может быть что угодно: число, строка, список, другой словарь |

**Важно:** списки и словари **не могут быть ключами** — они изменяемы. Строки, числа и кортежи — могут.

````python
# Пример 1. Создание словарей
english_to_russian = {"hello": "привет", "cat": "кот", "book": "книга"}
student_scores = {"Anna": 85, "Ivan": 72, "Olga": 91}
empty_dict = {}

print("Перевод:", english_to_russian)
print("Оценки:", student_scores)
print("Пустой:", empty_dict)

# Словарь из пар
pairs = dict([("a", 1), ("b", 2), ("c", 3)])
print("Из пар:", pairs)

# Словарь из двух списков одинаковой длины
keys = ["name", "age", "city"]
values = ["Anna", 25, "Moscow"]
person = dict(zip(keys, values))
print("Из zip:", person)
````

## 1.2. Доступ к элементам

### 1.2.1. Через `[]`

    d["hello"]   # 'привет'

Если ключа нет — возникнет `KeyError`.

### 1.2.2. Через `.get()`

`get(key, default)` возвращает значение, а если ключа нет — **не падает**, а возвращает `default` (по умолчанию `None`).

    d.get("hello")            # 'привет'
    d.get("unknown")          # None
    d.get("unknown", "—")     # '—'

**Правило:** если ключ может отсутствовать — используйте `get()`. Это самый частый способ избежать `KeyError`.

````python
# Пример 2. Доступ к элементам
d = {"hello": "привет", "cat": "кот", "book": "книга"}

# Через [] — безопасно, если ключ точно есть
print("d['hello'] =", d["hello"])

# Через .get() — безопасно всегда
print("d.get('cat') =", d.get("cat"))
print("d.get('unknown') =", d.get("unknown"))
print("d.get('unknown', '—') =", d.get("unknown", "—"))

# Проверка наличия ключа
print("'hello' in d:", "hello" in d)
print("'dog' in d:", "dog" in d)
````

## 1.3. Добавление, изменение и удаление элементов

### Добавление и изменение

    d["dog"] = "собака"       # добавит новый ключ
    d["cat"] = "кошка"        # изменит значение существующего

### Удаление

| Операция | Что делает |
|----------|------------|
| `del d[key]` | Удаляет ключ; `KeyError`, если ключа нет |
| `d.pop(key)` | Удаляет и возвращает значение |
| `d.pop(key, default)` | Удаляет, если есть, иначе возвращает `default` |
| `d.popitem()` | Удаляет и возвращает последнюю добавленную пару (в Python 3.7+) |
| `d.clear()` | Очищает словарь |

````python
# Пример 3. Изменение и удаление
d = {"hello": "привет", "cat": "кот"}

d["dog"] = "собака"
print("После добавления:", d)

d["cat"] = "кошка"
print("После изменения:", d)

del d["hello"]
print("После del:", d)

value = d.pop("dog")
print("pop вернул:", value)
print("После pop:", d)

# Безопасное удаление
removed = d.pop("unknown", None)
print("pop несуществующего:", removed)
````

## 1.4. Методы словаря

| Метод | Что возвращает |
|-------|----------------|
| `d.keys()` | Все ключи (специальное представление) |
| `d.values()` | Все значения |
| `d.items()` | Пары `(ключ, значение)` |
| `d.get(key, default)` | Значение или `default` |
| `d.update(other)` | Обновляет словарь парами из `other` |
| `d.setdefault(key, default)` | Возвращает значение; если ключа нет — создаёт его со значением `default` |
| `d.copy()` | Поверхностная копия |

### Метод `setdefault` — мощный инструмент

    freq = {}
    for word in ["ai", "tools", "ai", "help"]:
        freq.setdefault(word, 0)
        freq[word] += 1

Или короче:

    freq[word] = freq.get(word, 0) + 1

Оба варианта дают одинаковый результат — это стандартный приём подсчёта частот.

````python
# Пример 4. Методы словаря
d = {"hello": "привет", "cat": "кот", "book": "книга"}

print("keys:", list(d.keys()))
print("values:", list(d.values()))
print("items:", list(d.items()))

# update — объединяем с другим словарём
d.update({"dog": "собака", "cat": "кошка"})
print("После update:", d)

# setdefault — счётчик частоты
freq = {}
for word in ["ai", "tools", "ai", "help", "ai"]:
    freq.setdefault(word, 0)
    freq[word] += 1
print("Частоты:", freq)
````

## 1.5. Перебор словаря

    for key in d:                # только ключи
        print(key)

    for key in d.keys():         # то же самое
        print(key)

    for value in d.values():     # только значения
        print(value)

    for key, value in d.items(): # ключ + значение
        print(key, "→", value)

Перебор через `.items()` — самый удобный и частый способ.

````python
# Пример 5. Перебор словаря
d = {"hello": "привет", "cat": "кот", "book": "книга"}

print("Только ключи:")
for key in d:
    print(" ", key)

print("Только значения:")
for value in d.values():
    print(" ", value)

print("Пары:")
for key, value in d.items():
    print(f"  {key} → {value}")
````

## 1.6. Генераторы словарей

Как и списки, словари можно строить через генератор:

    {k: v for k, v in ...}

Примеры:

    squares = {x: x**2 for x in range(5)}        # {0:0, 1:1, 2:4, 3:9, 4:16}
    lengths = {w: len(w) for w in ["ai", "tools"]}  # {'ai': 2, 'tools': 5}

С фильтром:

    long_only = {w: len(w) for w in ["ai", "tools", "help"] if len(w) > 2}

Генераторы словарей — компактный и читаемый способ строить словари на основе последовательностей.

````python
# Пример 6. Генераторы словарей
squares = {x: x**2 for x in range(5)}
print("Квадраты:", squares)

words = ["ai", "tools", "help", "learn", "English"]
lengths = {w: len(w) for w in words}
print("Длины слов:", lengths)

long_only = {w: len(w) for w in words if len(w) > 3}
print("Только длинные:", long_only)

# Инвертирование словаря
d = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in d.items()}
print("Инвертированный:", inverted)
````

# Часть 2. Кортежи (`tuple`)

## 2.1. Что такое кортеж

**Кортеж** (`tuple`) — это **неизменяемая** упорядоченная последовательность элементов. По сути, это «замороженный» список: после создания его нельзя менять.

### 2.1.1. Синтаксис

    t1 = (1, 2, 3)
    t2 = 1, 2, 3            # скобки необязательны
    t3 = (5,)               # кортеж из одного элемента — нужна запятая
    t4 = ()                 # пустой кортеж

**Внимание:** `(5)` — это просто число 5, а `(5,)` — кортеж из одного элемента.

### 2.1.2. Когда использовать кортеж, а не список

| Ситуация | Что использовать |
|----------|------------------|
| Данные не должны меняться (например, координаты, дата) | Кортеж |
| Данные будут меняться | Список |
| Нужен ключ в словаре | Только кортеж |
| Нужен элемент множества | Только кортеж |
| Возврат нескольких значений из функции | Обычно кортеж |

Кортежи **быстрее** списков и **защищены от случайных изменений**.

````python
# Пример 7. Создание кортежей
t1 = (1, 2, 3)
t2 = 1, 2, 3
t3 = (5,)
t4 = ()

print("t1:", t1, "| тип:", type(t1).__name__)
print("t2:", t2, "| тип:", type(t2).__name__)
print("t3:", t3, "| тип:", type(t3).__name__)
print("t4:", t4, "| длина:", len(t4))

# Проверка: (5) — это число, (5,) — кортеж
print("type((5)) =", type((5)).__name__)
print("type((5,)) =", type((5,)).__name__)
````

## 2.2. Операции с кортежами

Работают все операции, не изменяющие содержимое:

- индексы: `t[0]`, `t[-1]`;
- срезы: `t[1:3]`;
- `len(t)`;
- `x in t`;
- `t + t2` (конкатенация);
- `t * 3` (повтор);
- `t.count(x)`, `t.index(x)`.

**Не работают** методы, изменяющие содержимое: `append`, `remove`, `sort`, `pop` и т. п.

````python
# Пример 8. Операции с кортежами
t = (10, 20, 30, 40, 50)

print("t[0] =", t[0])
print("t[-1] =", t[-1])
print("t[1:4] =", t[1:4])
print("len(t) =", len(t))
print("30 in t:", 30 in t)
print("t + (60, 70) =", t + (60, 70))
print("t * 2 =", t * 2)
print("t.count(20) =", t.count(20))
print("t.index(30) =", t.index(30))
````

## 2.3. Распаковка кортежей

Кортеж удобно «распаковать» в переменные:

    point = (3, 7)
    x, y = point

Работает и с любыми последовательностями:

    a, b, c = [1, 2, 3]
    first, *rest = [1, 2, 3, 4]     # first=1, rest=[2, 3, 4]

### Обмен значениями

    a, b = b, a

Это тоже распаковка — Python создаёт временный кортеж `(b, a)`.

````python
# Пример 9. Распаковка
point = (3, 7)
x, y = point
print(f"x = {x}, y = {y}")

a, b, c = [10, 20, 30]
print(f"a = {a}, b = {b}, c = {c}")

first, *rest = [1, 2, 3, 4]
print(f"first = {first}, rest = {rest}")

# Обмен
a, b = 1, 2
a, b = b, a
print(f"После обмена: a = {a}, b = {b}")
````

## 2.4. Кортежи как ключи словаря

Списки нельзя использовать как ключи словаря, а кортежи — можно. Это удобно, когда ключ состоит из нескольких частей:

    distances = {("Moscow", "SPb"): 705, ("Moscow", "Kazan"): 820}
    print(distances[("Moscow", "SPb")])   # 705

Так же кортежи можно класть в множества:
    points = {(0, 0), (1, 2), (3, 4)}

````python
# Пример 10. Кортежи в словарях и множествах
distances = {("Moscow", "SPb"): 705, ("Moscow", "Kazan"): 820}
print("Moscow → SPb:", distances[("Moscow", "SPb")], "км")

points = {(0, 0), (1, 2), (3, 4), (1, 2)}   # дубликат (1,2) исчезнет
print("Уникальные точки:", points)

# А вот список как ключ — вызовет ошибку
try:
    bad = {[1, 2]: "x"}
except TypeError as e:
    print("Ошибка:", e)
````

# Часть 3. Множества (`set`)

## 3.1. Что такое множество

**Множество** (`set`) — это **неупорядоченная** коллекция **уникальных** элементов. Порядок элементов не сохраняется и не гарантируется.

### 3.1.1. Синтаксис

    s1 = {1, 2, 3}
    s2 = set([1, 2, 2, 3, 3])   # {1, 2, 3} — дубликаты исчезли
    s3 = set()                  # пустое множество

**Внимание:** `{}` — это пустой **словарь**, а не множество! Пустое множество создаётся только через `set()`.

### 3.1.2. Когда использовать множества

- убрать дубликаты из списка;
- проверить, входит ли элемент в группу (быстрее, чем в списке);
- сравнивать наборы: какие элементы общие, какие только в одном наборе;
- посчитать количество уникальных слов в тексте.

````python
# Пример 11. Создание множеств и удаление дубликатов
s1 = {1, 2, 3, 3, 2, 1}
print("s1:", s1)     # {1, 2, 3}

words = ["ai", "tools", "ai", "help", "tools"]
unique = set(words)
print("Уникальные слова:", unique)
print("Количество уникальных:", len(unique))

print("Пустое множество:", set())
print("Пустой словарь:", {})
````

## 3.2. Операции над множествами

### 3.2.1. Добавление и удаление

| Метод | Что делает |
|-------|------------|
| `s.add(x)` | Добавляет элемент |
| `s.remove(x)` | Удаляет элемент; `KeyError`, если нет |
| `s.discard(x)` | Удаляет элемент, если есть; **не падает**, если нет |
| `s.pop()` | Удаляет и возвращает случайный элемент |
| `s.clear()` | Очищает множество |

### 3.2.2. Теоретико-множественные операции

Пусть `a = {1, 2, 3}`, `b = {3, 4, 5}`.

| Операция | Метод | Оператор | Результат |
|----------|-------|----------|-----------|
| Объединение | `a.union(b)` | `a | b` | `{1,2,3,4,5}` |
| Пересечение | `a.intersection(b)` | `a & b` | `{3}` |
| Разность | `a.difference(b)` | `a - b` | `{1,2}` |
| Симметричная разность | `a.symmetric_difference(b)` | `a ^ b` | `{1,2,4,5}` |

### 3.2.3. Проверки

| Проверка | Что означает |
|----------|---------------|
| `a.issubset(b)` или `a <= b` | Все элементы `a` входят в `b` |
| `a.issuperset(b)` или `a >= b` | Все элементы `b` входят в `a` |
| `a.isdisjoint(b)` | Множества не пересекаются |

````python
# Пример 12. Операции над множествами
a = {1, 2, 3}
b = {3, 4, 5}

print("Объединение:      ", a | b)
print("Пересечение:      ", a & b)
print("Разность a - b:   ", a - b)
print("Разность b - a:   ", b - a)
print("Симм. разность:   ", a ^ b)

print("issubset:", {1, 2}.issubset(a))
print("issuperset:", a.issuperset({1, 2}))
print("isdisjoint:", a.isdisjoint({10, 20}))
````

````python
# Пример 13. Применение множеств: уникальные слова в тексте
text1 = "ai tools help learn english"
text2 = "ai python help programming"

words1 = set(text1.split())
words2 = set(text2.split())

print("Слова первого текста:", sorted(words1))
print("Слова второго текста:", sorted(words2))
print("Общие слова:        ", sorted(words1 & words2))
print("Только в первом:    ", sorted(words1 - words2))
print("Только во втором:   ", sorted(words2 - words1))
````

# Часть 4. Модуль `collections`

Стандартная библиотека Python предоставляет модуль `collections`, в котором есть удобные специализированные структуры данных.

## 4.1. `Counter` — счётчик

`Counter` — это подкласс словаря, предназначенный для подсчёта. Он делает код короче и читаемее.

### Базовое использование

    from collections import Counter
    c = Counter("abracadabra")
    print(c)              # Counter({'a': 5, 'b': 2, 'r': 2, 'c': 1, 'd': 1})
    print(c.most_common(2))   # [('a', 5), ('b', 2)]

### Полезные методы

| Метод | Что делает |
|-------|------------|
| `c.most_common(n)` | Топ-N самых частых |
| `c.total()` | Сумма всех счётчиков |
| `c.elements()` | Итератор по элементам с учётом частот |
| `c1 + c2` | Сложение счётчиков |
| `c1 - c2` | Вычитание |

````python
# Пример 14. Counter
from collections import Counter

text = "ai tools help ai learn tools ai"
words = text.split()

c = Counter(words)
print("Counter:", c)
print("Топ-2:  ", c.most_common(2))
print("Всего:  ", c.total())

# Частоты букв
letters = Counter("abracadabra")
print("Буквы:", letters)
print("Топ-3:", letters.most_common(3))
````

## 4.2. `defaultdict` — словарь со значением по умолчанию

`defaultdict` автоматически создаёт значение для отсутствующего ключа. Это избавляет от ручной проверки `if key not in d`.

    from collections import defaultdict

    groups = defaultdict(list)
    groups["a"].append(1)     # ключ 'a' создаётся автоматически
    groups["a"].append(2)
    groups["b"].append(3)
    print(dict(groups))       # {'a': [1, 2], 'b': [3]}

    counts = defaultdict(int) # значение по умолчанию — 0
    for w in ["ai", "tools", "ai"]:
        counts[w] += 1
    print(dict(counts))       # {'ai': 2, 'tools': 1}

### Где это удобно

- группировка слов по первой букве;
- подсчёт частот без `setdefault`;
- накопление списков по ключу.

````python
# Пример 15. defaultdict — группировка слов по первой букве
from collections import defaultdict

words = ["apple", "apricot", "banana", "blueberry", "cherry", "avocado"]
by_letter = defaultdict(list)

for w in words:
    by_letter[w[0]].append(w)

for letter, group in sorted(by_letter.items()):
    print(letter, "→", group)
````

## 4.3. `namedtuple` — кортеж с именами полей

`namedtuple` — это кортеж, у которого у каждого элемента есть имя. Удобно для записей:

    from collections import namedtuple

    Student = namedtuple("Student", ["name", "score"])
    s = Student("Anna", 85)
    print(s.name, s.score)     # Anna 85
    print(s[0], s[1])           # Anna 85 — доступ по индексу тоже работает

### Преимущества

- читаемость: `s.name` вместо `s[0]`;
- неизменяемость, как у кортежа;
- работает везде, где работает кортеж (в т. ч. как ключ словаря).

````python
# Пример 16. namedtuple
from collections import namedtuple

Student = namedtuple("Student", ["name", "score", "level"])
students = [
    Student("Anna", 85, "B2"),
    Student("Ivan", 72, "B1"),
    Student("Olga", 91, "C1"),
]

for s in students:
    print(f"{s.name}: балл {s.score}, уровень {s.level}")

# Доступ по индексу тоже работает
print("Первый студент:", students[0][0])
````

# Часть 5. Типичные ошибки

| Ошибка | Что происходит | Как исправить |
|--------|----------------|----------------|
| `d["missing"]` | `KeyError` | Используйте `d.get("missing", default)` |
| Список как ключ: `{[1,2]: "x"}` | `TypeError: unhashable type` | Используйте кортеж: `{(1,2): "x"}` |
| `{}` и `set()` | `{}` — это словарь, не множество | Пустое множество — только `set()` |
| `t = (5)` — не кортеж | Это число 5 | Нужна запятая: `t = (5,)` |
| `s.add([1,2])` | `TypeError` — список не hashable | Добавляйте кортежи |
| Изменение словаря во время перебора | `RuntimeError: dictionary changed size` | Перебирайте `list(d.keys())` или собирайте изменения отдельно |
| `d1 = d2` и изменение `d1` | Изменяется и `d2` (это одна ссылка) | Копируйте: `d1 = d2.copy()` или `dict(d2)` |
| Порядок элементов множества | Может отличаться от ожидаемого | Не полагайтесь на порядок; для порядка — сортируйте |
| `Counter` считает по элементам | Порядок в `most_common` нестабилен при равных частотах | При равных частотах порядок может отличаться между запусками |
